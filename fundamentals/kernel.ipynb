```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from scipy.ndimage import gaussian_filter
from skimage import io
```

```python
image = io.imread("./images/picture_dog.jpg", as_gray=True)
image_suave = gaussian_filter(image, 2)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(image, cmap="gray")
ax[1].imshow(image_suave, cmap="gray")

for a in ax:
    a.axis("off")
plt.show()
```

```python
laplacian = np.array([[0,  1, 0],
                      [1, -4, 1],
                      [0,  1, 0]])
```

```python
edges = np.abs(convolve2d(image, laplacian, mode="same", boundary="symm"))
edges_suave = np.abs(convolve2d(image_suave, laplacian, mode="same", boundary="symm"))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

ax[0].imshow(edges, cmap="gray_r", vmax=np.percentile(edges, 99))
ax[1].imshow(edges_suave, cmap="gray_r", vmax=np.percentile(edges_suave, 99))

for a in ax:
    a.axis("off")

plt.show()
```
